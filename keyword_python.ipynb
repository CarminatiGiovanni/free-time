# global
refers to the correct namespace

```python
name = 'Giovanni'

def change_name_wrong():
    name = 'Nicole'

def change_name():
    global name
    name = 'Pri'

print(name)
change_name_wrong()
print(name)
change_name()
print(name)
```

# lambda

```python
from typing import Callable, Any
p = lambda x: print(x)

p('ciao')

(lambda x: print(x))('altro')

def apply_result(f: Callable, item:list[Any]):
    for i in item:
        i = f(i)
        print(i)

apply_result(lambda x: x**2, [c for c in range(1,5)])
```

# yield, next

```python
# used for generator

from typing import Generator

def generate_numeri(limit: int) -> Generator:
    for i in range(0,limit):
        yield i

numb: int = generate_numeri(9)
print(numb)
print(next(numb))
print(next(numb))
print(next(numb))
print(next(numb))
print(list(numb))

print(type(range(0,4)))
print(type(next))
```
