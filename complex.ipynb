```python
import numpy as np
import matplotlib.pyplot as plt

def setup_complex_plane(ax, title):
    # Set limits
    ax.set_xlim(-2, 2)
    ax.set_ylim(-2, 2)

    # Move axes to center
    ax.spines['left'].set_position('center')
    ax.spines['bottom'].set_position('center')

    # Hide top and right spines
    ax.spines['right'].set_color('none')
    ax.spines['top'].set_color('none')

    # Set ticks position
    ax.xaxis.set_ticks_position('bottom')
    ax.yaxis.set_ticks_position('left')

    # Grid
    ax.grid(True, linestyle='--', alpha=0.6)

    # Labels
    #ax.set_xlabel('Re')
    #ax.set_ylabel('Im')
    ax.set_title(title)


# Create figure with two subplots
fig, axs = plt.subplots(1, 2, figsize=(12, 6))

# First complex plane
setup_complex_plane(axs[0], "z")

# Example points in first plane
Φ1 = 60/180 * np.pi 
ρ = np.linspace(0,np.sqrt(2),100)
Φ2 = 30/180 * np.pi
z1 = ρ*np.exp(1j*Φ1)
z2 = ρ*np.exp(1j*Φ2)
axs[0].plot(z1.real, z1.imag, 'b-')
axs[0].plot(z2.real, z2.imag, 'b-')
# axs[0].text(z1.real, z1.imag, ' 2+3j', color='red')

# Second complex plane
setup_complex_plane(axs[1], "f(z)")

# Example points in second plane
z2 = -1 + 2j
axs[1].plot(z2.real, z2.imag, '-')
# axs[1].text(z2.real, z2.imag, ' -1+2j', color='blue')

plt.tight_layout()
plt.show()
```
